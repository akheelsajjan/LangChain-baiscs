```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
print(os.getenv("OPENAI_API_KEY"))
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-3.5-turbo")
```

```python
from langchain.messages import SystemMessage, HumanMessage

system_message = SystemMessage("""You are a senior Python developer.
 Provide examples of how to use the langchain library to create agents.
""")

message = [
    system_message,
    HumanMessage("how to create an agent using langchain?")
]

```

```python
response =model.invoke(message)
print(response.content)
```

```python
from langchain.messages import AIMessage, ToolMessage

ai_message = AIMessage(
    content="The current weather in Dharwad is 30 degrees Celsius.",
    tool_calls=[
            {
                'name': "check_weather",
                
                'args': {"temperature_unit": "Celsius"},
                'id': "tool_1"
            }
    ]
)

tool_message = ToolMessage(
    content="The current weather in Dharwad is 30 degrees Celsius.",
    tool_call_id="tool_1"
)

message = [
    HumanMessage("what is the weather in Dharwad?"),
    ai_message,
    tool_message
]

response =model.invoke(message)
print(response.content)
```
