```python
import langchain
langchain__version__ = langchain.__version__
print(langchain__version__)
```

```python
from langchain.agents import create_agent

def check_weather(location: str) -> str:
    '''Return the weather forecast for the specified location.'''
    return f"It's always sunny in {location}"

agent = create_agent(
    model="gpt-3.5-turbo",
    tools=[check_weather],
    system_prompt="You are a helpful assistant",
)

```

```python
agent.invoke({"messages": [{"role": "user", "content": "what is the weather in Dharwad?"}]})
```
