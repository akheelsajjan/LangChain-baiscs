STREAMING : models can stream their outpus while response is genereated

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
print(os.getenv("OPENAI_API_KEY"))
```

```python
from langchain_openai import ChatOpenAI

model = ChatOpenAI(model="gpt-3.5-turbo")
```

```python
for chunk in model.stream("Write 200 words about India"):
    print(chunk.text, end="|" , flush=True)
```

BATCH : collection of Independent request 

```python
model.batch([
    "Write 50 words about India", 
    "Write 50 words about USA",
    "Write 50 words about China"],
    config={
        'max_concurrency': 4
    }
    )
```
